```python
import pandas as pd

df = pd.DataFrame({
    "Region": ["East", "West", "North", "South", "East", "West", "North", "South"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile"],
    "Sales": [55000, 30000, 24000, 66000, 45000, 36000, 88000, 27000],
    "Quantity": [5, 10, 8, 6, 15, 12, 8, 9]
})

print(df)
```

```python
resulted_colums=df.groupby("Region").agg({"Sales":["sum","mean"],"Quantity":["sum","max"]})
resulted_colums
```

```python
columns=["Region","sales_sum","sales_mean","quantity_sum","quantity_max"]
resulted_colums=columns
```
